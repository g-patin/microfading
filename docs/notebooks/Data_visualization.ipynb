# Import statements

# Retrieve files 

```python
# For didactic purposes, I am using the dataset files provided by the microfading package

files = mf.get_datasets(BWS=True)
```

# Basic plotting 

```python
m = mf.MFT(files=files)

d = m.plot_bwse()
```

# Advanced plotting 

For most plotting functions, there are one or two parameters for which users can directly enter their arguments. For instance, for the `plot_sp()` function, the parameters *spectra* and *stdev* can quickly be adjusted. Afterwards, the functions work with dictionaries to adjust some aspects of the figure (see code below). 

```python
m = mf.MFT(files=files)

m.plot_sp(
    spectra=('He',1),
    data_settings={
        'mode':'A',               # plot the absorbance
        'derivation': True,       
        'smoothing': (10,1),      # smooth the data, I usually leave the second number equal to 1
        'wl_range': (420,700)
    },
    figure_settings={
        'figsize':(15,6),
        'fontsize':18,
        'title':'MFT data'
    },    
    legend_settings={
        'labels':'default',
        'position': 'out',
        'fontsize':14,
        'title':"Meas Id"
    },    
    lines_settings={
        'colors':'sample',     
        'ls':'random',            # when random, it creates a bit of variation in the linestyles
        'lw':[2,1,3,1,4] 
    }
)
```

```python
m1 = mf.MFT(files=files[0:1])

df = m1.plot_sp(
    spectra=('Hv', [0,0.5,1]),
    data_settings={
        'mode':'A',               # plot the absorbance
        'derivation': False,       
        'smoothing': (10,1),      # smooth the data, I usually leave the second number equal to 1
        'wl_range': (420,700)
    },
    figure_settings={
        'figsize':(15,6),
        'fontsize':18,
        'title':'MFT data'
    },    
    legend_settings={
        'labels':'default',
        'position': 'out',
        'fontsize':14,        
    },    
    lines_settings={
        'colors':'k', #'sample',     
        'ls':'random',            # when random, it creates a bit of variation in the linestyles         
    }
)
```
