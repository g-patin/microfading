# Process rawfiles 

## basic functionalities

This is the most basic and straightforward way to process the microfading rawdata files. Only two arguments needs to be defined:

1. The paths of the raw files
2. The name of the microfading device

As a result, it produces an excel file for each analysis, which is saved in the current working directory of this notebook. 

```python
mf.process_rawdata(
    files=rawfiles,
    device='Fotonowy'    
)
```

```python
mf.process_rawdata(
    files=rawfiles,
    device='Fotonowy',
    db=False,
    white_standard='default',
    delete_files=False
)
```
